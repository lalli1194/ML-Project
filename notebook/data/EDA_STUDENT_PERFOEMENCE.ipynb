#### BIVARIATE ANALYSIS ( Is parental education has any impact on student's performance ? )


```python
import matplotlib.pyplot as plt
import pandas as pd

# Convert score columns to numeric (safe step)
score_cols = ['math score', 'reading score', 'writing score']

for col in score_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Create average column if it doesn't exist
df['average'] = df[score_cols].mean(axis=1)

# Group by parental education and calculate mean only for numeric score columns
education_group = df.groupby('parental level of education')[
    ['math score', 'reading score', 'writing score', 'average']
].mean()

# Plot
education_group.plot(
    kind='barh',
    figsize=(10, 10)
)

plt.title('Average Scores by Parental Level of Education')
plt.xlabel('Average Marks')
plt.ylabel('Parental Level of Education')
plt.legend(title='Subjects', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

#### Insights 
- The score of student whose parents possess master and bachelor level education are higher than others
